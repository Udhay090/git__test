````python
# simple countdown
count=5
while count > 0:
    print(count)
    count -= 1 #important:update the condition
    print("Blastoff!")
````

````python
#countdown 1 to 10
num=1
while num<=10 :
    print(num)
    num+=1
````

````python
#skip multiples of 3
for i in range(1,11):
   if i % 3==0:
     continue#
   print(i,end=" ")

print("\nmultiples of 3 were skipped")
````
