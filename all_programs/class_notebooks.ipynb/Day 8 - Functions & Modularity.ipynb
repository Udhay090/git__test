## 1. Why Functions?

Functions help you:
- Avoid repeating code (DRY [Dont' Repeat Yourself]principle)
- Organize logic into small, testable units
- Improve readability and reuse


## 2. Defining and Calling Functions

### Syntax:
```python
def name(parameters):
    # body
    return value
```


````python
def greet(name):
    print(f'Hello, {name}!')

greet('Alice')
greet(1)
````

## 3. Return Values


````python
def add_and_subtract(a, b):
    result = a + b
    res2 = a - b
    return [result,res2]

sum_val = add_and_subtract(3,5)
print('3 + 5 =', sum_val)
````

## 4. Default and Keyword Arguments


````python
def power(base = 2, exponent = 2):
    return base ** exponent

exponent = 3
print(power(3,exponent))      # 3^2
print(power(3, 3))   # 3^3
print(power(base=2, exponent=5))
````

## 5. Variable Scope

- Variables inside functions are **local**
- Global variables exist outside functions


````python
x = 10  # global

def modify():
    x = 20  # local
    print('Inside:', x)

modify()
print('Outside:', x)
````

## 6. Utility Functions
Let's build small reusable functions.


````python
def bmi(weight, height):
    return weight / (height ** 2)

def grade_from_percentage(p):
    if p >= 90:
      return 'A'
    elif p >= 80:
      return 'B'
    elif p >= 70:
      return 'C'
    elif p >= 60:
      return 'D'
    elif p >= 40:
      return 'E'
    else:
      return 'F'

print('BMI(70kg,1.75m)=', bmi(70,1.75))
print('Grade for 85% =', grade_from_percentage(85))
````
