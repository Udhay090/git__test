# 3. Downloading a Webpage

We first download the webpage using the **requests** library.


````python
url = "http://quotes.toscrape.com"
response = requests.get(url)

print(response.text)

# print("Status Code:", response.status_code)
````

If the status code is **200**, the request was successful.


# 4. Parse HTML with BeautifulSoup

BeautifulSoup converts HTML text into a structure that Python can navigate.


````python
soup = BeautifulSoup(response.text, "html.parser")

print(soup.title)
print(soup.title.text)
````

# 5. Extract Quotes from Page

Each quote is inside:

```html
<span class="text">
```


````python
quotes = soup.find_all("span", class_="text")

for q in quotes:
    print(q.text)
````

# Exercise 2 – Extract Authors

Each author name appears in:

```html
<small class="author">
```

Write code to extract authors.


````python
authors = soup.find_all("small", class_="author")

for a in authors:
    print(a.text)
````

# 6. Extract Multiple Fields

We want structured data like:

| quote | author |


````python
quote_blocks = soup.find_all("div", class_="quote")

data = []

for block in quote_blocks:

    quote = block.find("span", class_="text").text
    author = block.find("small", class_="author").text

    data.append({
        "quote": quote,
        "author": author
    })

data[:5]
````

# 7. Convert Data into Dataset


````python
df = pd.DataFrame(data)
df.head()
````

````python
df.to_csv("quotes_dataset.csv", index=False)
````

Now we have created our **first dataset from the internet**.

# 8. Scraping Multiple Pages

Many websites contain **pagination**.

Example:

http://quotes.toscrape.com/page/1  
http://quotes.toscrape.com/page/2


````python
all_data = []

for page in range(1,6):

    url = f"http://quotes.toscrape.com/page/{page}/"

    url = "http://quotes.toscrape.com/page/1/"
    url = "http://quotes.toscrape.com/page/2/"

    response = requests.get(url)
    soup = BeautifulSoup(response.text, "html.parser")

    quotes = soup.find_all("div", class_="quote")

    for q in quotes:

        quote = q.find("span", class_="text").text
        author = q.find("small", class_="author").text

        all_data.append({
            "quote": quote,
            "author": author
        })

df = pd.DataFrame(all_data)

print(df)
````

````python
df.to_csv("quotes_full_dataset.csv", index=False)
````

# 9. Example 2 – Scraping Books Dataset

Website:

http://books.toscrape.com

We will scrape:

- Book title
- Price


````python
url = "http://books.toscrape.com"

response = requests.get(url)
soup = BeautifulSoup(response.text,"html.parser")

books = soup.find_all("article", class_="product_pod")

books_data = []

for book in books:

    title = book.h3.a["title"]
    price = book.find("p", class_="price_color").text

    books_data.append({
        "title": title,
        "price": price
    })

books_df = pd.DataFrame(books_data)

books_df.head()
````

# Exercise 4 – Save Books Dataset

````python
books_df.to_csv("books_dataset.csv", index=False)
````

# 11. Example 3 – Scraping Job Listings

Dataset fields:

- Job Title
- Company
- Location

Website:

https://realpython.github.io/fake-jobs/


````python
url = "https://realpython.github.io/fake-jobs/"

response = requests.get(url)
soup = BeautifulSoup(response.text,"html.parser")

jobs = soup.find_all("div", class_="card-content")

job_data = []

for job in jobs:

    title = job.find("h2").text.strip()
    company = job.find("h3").text.strip()
    location = job.find("p", class_="location").text.strip()

    job_data.append({
        "title": title,
        "company": company,
        "location": location
    })

jobs_df = pd.DataFrame(job_data)

jobs_df.head()
````
