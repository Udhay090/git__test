```python
print("pattern 1 : right triangle")
for i in range (1,5):#range (1,6) means 0,1,2,3,4,5
   for j in range(i):
      print("*",end="")
   print( )
```

```python
print("pattern 2 : inverted triangle")
for i in range (6,0,-1):#range (1,6) means 0,1,2,3,4,5
   for j in range(i):
      print("*",end="")
   print( )
```

```python
   # pattern 3
print("pattern 3: pyramid")
n=5
for i in range(1, n+1):
     for j in range(n-i):# it prints n-i spaces
           print(" ",end="")
     for k in range (2 * i - 1):#it prints 2i-1 stars in ith row
      print("*",end="")
     print()
```

6.list comprehensions(intro)

a list comprehension is a concise way to create lists.


```python
squares=[x**2 for x in range(1,6)]
print(squares)
print("even squares:",[x**2 for x in range(1,11) if x%2==0])
              
```
