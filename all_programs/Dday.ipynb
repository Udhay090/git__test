```python
# the points that are printed are indexed as (i,j) and in 2d plane and as long as 
#they are indexed with the condition that i>=j we get a right triangle or a left shifted pyramid
n=int(input("enter the value of n :"))
for i in range(n):
    for j in range(n):
        if i >= j:
            print("*",end=" ")# end statement helps you to print next line after the 1st line  
        else:
            print("",end=" ")
    print()            
 
```

```python
#inverted pascals triangle
n=int(input("enter the value of n :"))
for i in range(n+1):
    for j in range(n+1):
        if i>=j:
            print(" ",end=" ")# end statement helps you to print next line after the 1st line  
        else:
            print("*",end=" ")
    print()            
 
```

```python
# here we have done -ve indexing for j axis from -
n=int(input("enter the value of n :"))
for i in range(n):
    for j in range(-n,n):
        if -i<=j and j<=i:
            print("*",end=" ")
        else:
            print(" ",end=" ")
    print()            
 
```

```python
# here we have done -ve indexing for j axis from -
# bookmark about to print
n=int(input("enter the value of n :"))
for i in range(n):
    for j in range(-n,n+1):
        if -i<=j and j<=i:
            print(" ",end=" ")
        else:
            print("*",end=" ")
    print()            
 
```
