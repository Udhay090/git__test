## 4. Methods in a Class

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def display(self):
        print(f'Name: {self.name}, Marks: {self.marks}')

s = Student('Rahul', 85)
s.display()
```

## 6. Encapsulation

Encapsulation means binding data and methods together.
It also involves restricting direct access to some variables.

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance  # private variable

    def deposit(self, amount):
        self.__balance += amount

    def get_balance(self):
        return self.__balance

acc = BankAccount(1000)
acc.deposit(500)
print(acc.get_balance())
```

## 7. Inheritance

Inheritance allows a class to acquire properties of another class.

```python
class Person:
    def __init__(self, name):
        self.name = name

    def greet(self):
        print('Hello, my name is', self.name)

class Student(Person):
    def __init__(self, name, roll):
        super().__init__(name)
        self.roll = roll

s = Student('Ankit', 101)
s.greet()
```

## 8. Polymorphism

Polymorphism means same method name but different behavior.

```python
class Dog:
    def speak(self):
        print('Bark')

class Cat:
    def speak(self):
        print('Meow')

for animal in (Dog(), Cat()):
    animal.speak()
```

## 9. OOP in Data Science Context

- ML models are classes
- Data pipelines use objects
- Libraries like scikit-learn are fully OOP-based

```python
class Dataset:
    def __init__(self, data):
        self.data = data

    def mean(self):
        return sum(self.data) / len(self.data)

d = Dataset([10, 20, 30])
print(d.mean())
```
