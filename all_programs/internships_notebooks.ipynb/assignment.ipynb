16. Create a program to check if a number is prime using a function. 

```python
def is_prime(n):
    if n < 2: return False
    for i in range(2, int(n**0.5)+1):
        if n % i == 0: return False
    return True

print(is_prime(101))
```

24. Create a modular program for a library system (add books, issue books, return books).

```python
books=[]
while True:
    op=input("add/issue/return/end: ")
    if op=="end": break

    if op=="add":
        b=input("book: "); books.append(b)

    elif op=="issue":
        b=input("book: ")
        if b in books: books.remove(b)

    elif op=="return":
        b=input("book: "); books.append(b)

print("Books:", books)
```
