1.Dictionary basics
a dictionary stores data as key-value pairs
Syntax:
student={'name':'Alice','age':20} 

```python
student={'name':'Alice','age':20,'marks':88}#key value pairs, key acts like an index
print("student:",student)
#access values
print("name",student['name'])
print("age:",student['age'])
```

```python
print(student.get('name'))
print(student.get('name'))
print(student.get('age'))
```

```python
#.iterating over dictonary
print('keys:')
for k in student.keys():
    print(k)
print('values:')
for v in student.values():
    print(v)
    print('Items:')
for k,v in student.items():
    print(k,'>',v)
```

7.sets
A set is an unordered collection of unique elements 

```python
nums=[1,2,3,3,3,4,4]
s=set(nums)
print('original:', nums )
print('after removing duplicates:', s)
s.add(5)
print('after adding 5:', s)
s.remove(3)
print('after removing 3:', s)
print('is 4 present?',4 in s)
print('is 3 present?',3 in s)
```

```python
s={1,2,3,4,}
print(s)
s.add(8)
print(s)
```

```python
a={1,2,3,4}#set 
b={3,4,5,6}
print("a:", a)
print("b:", b)
print("Union of a and b:", a|b)
print("Intersection of a and b:", a-b )
print("Difference of a and b:", a.difference(b))
print("Symmetric difference of a and b:", a.symmetric_difference(b))
```

4.Default and keyword arguments

```python
def power(base,exponent=2):#default argument coz exponent=2
    return base**exponent
print(power(9,4))
print(power(base=2,exponent=5))#keyword argument coz we gave value 5 to exponent
```

6.utility functions


```python
def bmi(weight,height):
    return weight/(height**2)
print("bmi is :",bmi(73,178))
def grade(p):
    if p>=90:
        return 'A'
    elif p>=80:
        return 'B'
    elif p>=70:
        return 'C'
    elif p>=60:
        return 'D'
    elif p>=50:
        return 'E'
    else:
        return 'F'
print("grade is:",grade(91)) 
```
