```python
class student:
    def __init__(self,name,marks):
        self.name=name
        self.marks=marks
       
    def result(self):
        return self.name
student1=student("Om",95)
student2=student("rishi",78) 
print(student2.result())
print(student2.marks)
print(student1.result())
```

```python
class car:
    def __init__(self,brand,price):
        self.brand=brand
        self.price=price
    def drive(self):
        print("the car is driving")
car1=car("bugatti",40000)
car2=car("maruti",2000)

print(car1.brand)
print(car2.price)
car1.drive()
print(car2.drive())
```

datahiding we use self.__attribute to make the data private 

```python
class bankaccount:
    def __init__(self,name,account_no,balance):
        self.name=name
        self.account_no=account_no
        self.balance=balance
    def check(self):
        return self.balance
a1=bankaccount("ram","5678",40000)
a2=bankaccount("shyam","2345",50000)
print(a2.name)
print(a1.check())
print(a2.check())
print(a2.account_no)
```

list,dictionary,set comprehension

```python
x=[1,2,3,4,6,7,8,9,10]
y=[i**2 for i in x if i%2==0]#list comp
print(y)
z=[i for i in x if i%2==0]#list comp
print(z)
dict={i:i**2 for i in x if i%2==0}#dictionary comp
print(dict)
set={i for i in x if i%2==0}# set comp
print(set)
a=[1,3,2323,1231,54,64,0,234]
a.sort()
print(a)
```
